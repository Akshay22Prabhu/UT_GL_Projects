## Q22. What is proportion of trips that is Business and what is the proportion of trips that is Personal? (3 points)

```python
Total_trip = uber['CATEGORY*'].dropna().count()
No_of_Business_trip = uber[uber['CATEGORY*']=='Business'].count()['CATEGORY*']
No_of_Personal_trip = uber[uber['CATEGORY*']=='Personal'].count()['CATEGORY*']
Proportion_of_Business_category = No_of_Business_trip/Total_trip
Proportion_of_Personal_category = No_of_Personal_trip/Total_trip
print('Proportion of trips that is Business is : ', Proportion_of_Business_category)
print('Proportion of trips that is Personal is : ', Proportion_of_Personal_category)
```
